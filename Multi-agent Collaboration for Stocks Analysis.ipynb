```python
from crewai_tools import ScrapeWebsiteTool, SerperDevTool
from crewai import Agent, Task, Crew
search_tool=SerperDevTool()
scrape_tool=ScrapeWebsiteTool()
```

```python
stock_researcher_agent=Agent(role='Stock Researcher'
,goal='Look up stock information on websites pertaining to mid and large cap stocks that have immense potential to grow in value based on the company initiatives, opportunities, competitor market etc.'
,backstory="""Specializing in financial markets, this agent  looks up stock market websites for key information on stocks that have growth potential in the coming 5 years.
This agent pulls up a list of stocks that have significant growth potential. This agent has sound financial knowledge and exceptional skills to search websites and extract valuable insights
"""
,verbose=True
,allow_delegation=True
,tools=[scrape_tool, search_tool]
)
```

```python
stock_valuer_agent=Agent(role='Stocks Valuation Expert'
, goal='Based on the list of stocks shared by the Stocks Researcher, look up the company financials and other useful metrics just as a financial analyst would and come up with a list of undervalued stocks '
,backstory="""Equipped with a deep understanding of financial
              markets ,  quantitative analysis and deep knowledge in stock valuation ,this agent
              analyses balance sheets, cashflow statements and income statements of these list of companies shared by the Stocks Researcher
              and identifies undervalued stocks""",
    verbose=True,
    allow_delegation=True,
    tools = [scrape_tool, search_tool]


)
```

```python
stock_researcher_task=Task(description=("""
Gather data from stock market websites of stocks of mid and large cap growth companies that are likely to grow in the next 5 years. These are stocks of companies who have a certain advantage or are into researching something groundbreaking and are ahead of competition
and have high probability of growth
""")
, expected_output=("""List of 20 such identified mid and large stocks with the ticker symbol as per NASDAQ, description of what the  company is into and the rationale for picking up this stock as growth stock""")
,agent=stock_researcher_agent
                        )
```

```python
stock_valuation_task=Task(description=("""
Analyse the financial data from balance sheet, income statement , cashflow statement and other publicly available financial documents and compute the financial ratios and based on the insights shared by the
analyst, come up with the top under-valued  stock recommendations

""")
,expected_output=("""A list of undervalued stocks, rationale in terms of key financial metrics and growth plans that helped come up with this recommendation. Add any financial ratios analysed to arrive at the result.
Provide details of the financial ratios found for the company.
""")
,agent=stock_valuer_agent,
                               )
```

```python
from crewai import Crew, Process
from langchain_openai import ChatOpenAI

financial_trading_crew=Crew(   agents=[stock_researcher_agent,
            stock_valuer_agent
            # ,execution_agent,
            # risk_management_agent
                                       ],

    tasks=[stock_researcher_task,
           stock_valuation_task,
          #  execution_planning_task,
          #  risk_assessment_task
           ],
    manager_llm=ChatOpenAI(model='gpt-3.5-turbo'
    , temperature=0.7),
    process=Process.hierarchical
    # , verbose=True
    )
```

```python
# Example data for kicking off the process
financial_trading_inputs = {
    'stock_selection': 'PLTR',
    'initial_capital': '1000',
    'risk_tolerance': 'Medium',
    'trading_strategy_preference': 'Short to Long Term Trading',
    'news_impact_consideration': True
}
```

```python
result = financial_trading_crew.kickoff()
```
